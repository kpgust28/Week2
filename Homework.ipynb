Question 9: At each stage, check if they are providing numbers or not. If not, tell them to do so.

```python
loop: bool = True
while(loop):
    user_age: str = input("What is your age?")
    if(user_age.isdecimal()):
        loop = False

if(int(user_age) == 200):
    print("You are a robot")
elif(int(user_age) >= 16):
    print("You can drive if you have your license")
elif(int(user_age) == 2):
    print("You have barely started walking")
else:
    print("You cannot drive or get your license")
```
